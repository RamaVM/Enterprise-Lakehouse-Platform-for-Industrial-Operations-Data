```python
from pyspark.sql import functions as F

maintenance_path = "s3://enterprise-lakehouse-data/silver/maintenance_snapshots/"
curated_path = "s3://enterprise-lakehouse-data/silver_curated/maintenance_events/"

silver_m_df = spark.read.parquet(maintenance_path)
```

Remove exact duplicates inside Silver

```python
bootstrap_df = (
    silver_m_df
    .dropDuplicates(["maintenance_id", "maintenance_event_hash"])
)
```

Write Curated (Bootstrap)

```python
(
    bootstrap_df
    .write
    .mode("overwrite")   # allowed ONLY for bootstrap
    .parquet(curated_path)
)
```
